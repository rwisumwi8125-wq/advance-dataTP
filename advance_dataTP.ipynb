```python
##create a list of 10 numbeers as per your choice

num = [1,4,2,6,9,7,2,8,5,4]
print("original list",num)

##add your name in the last of the list

num.append("Rosse")
print("after append", num)
##remove the element at 3rd position starting from beginning

remove_from_start = num.pop(2)
print("remove (3rd from start):", remove_from_start)
print("list now:", num)
##get the total number of elements in the list
print( "total number of elements now:",len(num))
```

```python
## create a list of 15 random numbers but one of the number will be 100

numbers = [2,3,1,7,100,16,14,8,9,4,10,12,3,6,11]
print( "Numbers:", numbers)

## insert 100 at 5th position

numbers.insert(5,100)
print("After insert:", numbers)

## sort the new list without changing the original one

sorted_num = sorted(numbers)
print("Original:", numbers)
print( "sorted:",sorted_num)
```
